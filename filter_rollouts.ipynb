```python
from datasets import load_dataset, Dataset
import sys
sys.path.append("/user/int")
from grader import compute_score
import os, pickle
from tqdm import tqdm
```

```python
dataset = load_dataset("CMU-AIRe/InT-hard-set", split="train")
```

### Load student solutions

```python
results_dir = "OUTPUT_PATH"
all_results = {}

for start in tqdm(range(0, len(dataset), 512)):
    end = min(start + 512, len(dataset))
    pkl_name = f"pass_at_4_{start}_{end}_train.pkl"
    if os.path.exists(os.path.join(results_dir, pkl_name)):
        pkl_path = os.path.join(results_dir, pkl_name)
        with open(pkl_path, "rb") as f:
            part = pickle.load(f)
        all_results.update(part)
    else:
        print(f"Skipping {start // 512} because it does not exist.")

print(f"Loaded {len(all_results)} results.")
```

```python
rows = []
for i in tqdm(range(len(dataset))):
    student_solutions = [all_results[i].outputs[j].text for j in range(4)]
    
    for sol in student_solutions:
        if compute_score(sol, dataset[i]['answer']) == 1:
            continue
        row = dict(dataset[i])  # Make a new row for each student solution
        row['incorrect_attempt'] = sol
        rows.append(row)

dataset_with_student_solutions = Dataset.from_list(rows)
```

```python
dataset_with_student_solutions = dataset_with_student_solutions.filter(lambda x: x['solution'] != '')
dataset_with_student_solutions = dataset_with_student_solutions.filter(lambda x: x['solution'] is not None)
```

```python
dataset_with_student_solutions.push_to_hub("CMU-AIRe/InT-hard-set-with-incorrect-attempts", split="train")
```
